## 3. Zoek één variabele tegelijk

De functie zoekt alleen in de beschrijving en conceptnaam, en toont relevante kolommen. Zelf de juiste regels selecteren en dus concept id


```python
def zoek_variabele(zoekterm, max_regels=150):
    kolommen = ['source_code', 'source_code_description', 'concept_id',
                'concept_name', 'domain_id', 'source_vocabulary_id']
    bestaande = [k for k in kolommen if k in dictionary.columns]
    naam = dictionary.get('concept_name', pd.Series('', index=dictionary.index)).fillna('').astype(str)
    bron = dictionary.get('source_code_description', pd.Series('', index=dictionary.index)).fillna('').astype(str)
    gevonden = dictionary[naam.str.contains(zoekterm, case=False, na=False, regex=True)
                           | bron.str.contains(zoekterm, case=False, na=False, regex=True)]
    gevonden = gevonden[bestaande].drop_duplicates()
    print(f'{len(gevonden)} mogelijke regels voor {zoekterm!r}; hier zie je maximaal {max_regels}.')
    display(gevonden.head(max_regels))
    return gevonden
```
